# Python While Loop

This notebook explains **while loops in Python** with clear definitions, explanations, examples, use cases, and practice problems.

## 1. Definition

A **while loop** in Python repeatedly executes a block of code **as long as a given condition is True**.

## 2. Explanation

- A while loop is used when the **number of iterations is not known in advance**.
- The loop continues until the condition becomes False.
- You must update the condition inside the loop to avoid an infinite loop.

**Syntax:**
```python
while condition:
    # code block
```

## 3. Basic Examples

````
i = 1
while i <= 5:
    print(i)
    i += 1
````

````
count = 5
while count > 0:
    print(count)
    count -= 1
````

````
i = 0
while i < len("Python"):
    print("Python"[i])
    i += 1
````

````
num = 2
while num <= 20:
    print(num)
    num += 2
````

````
i = 1
while i <= 5:
    print(i * i)
    i += 1
````

## 4. Use Cases of While Loop

- When the number of iterations is unknown
- Reading input until a condition is met
- Menu-driven programs
- Games and simulations
- Continuous monitoring systems

## 5. While Loop with Conditions

````
i = 1
while i <= 10:
    if i % 2 == 0:
        print(i, "is even")
    i += 1
````

````
numbers = [3, 7, 12, 5, 18]
i = 0
while i < len(numbers):
    if numbers[i] > 10:
        print(numbers[i])
    i += 1
````

````
i = 1
while i <= 5:
    if i == 3:
        i += 1
        continue
    print(i)
    i += 1
````

````
i = 1
while i <= 5:
    if i == 4:
        break
    print(i)
    i += 1
````

````
while True:
    user_input = "exit"
    if user_input == "exit":
        print("Loop exited")
        break
````
