# Python For Loop

This notebook explains **for loops in Python** with definitions, explanations, examples, use cases, and practice problems.

## 1. Definition

A **for loop** in Python is used to iterate over a sequence (like a list, tuple, string, or range) and execute a block of code for each element in that sequence.

## 2. Explanation

- A for loop repeats a block of code **a fixed number of times**.
- It works with **iterable objects**.
- Python handles the loop counter automatically.

**Syntax:**
```python
for variable in sequence:
    # code block
```

## 3. Basic Examples

````
for i in range(5):
    print(i)
````

````
name = "Python"
for ch in name:
    print(ch)
````

````
numbers = [10, 20, 30, 40]
for num in numbers:
    print(num)
````

````
for i in range(1, 6):
    print(i * i)
````

````
fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit.upper())
````

## 4. Use Cases of For Loop

- Traversing lists, tuples, sets, dictionaries
- Repeating tasks with known iterations
- Processing data in machine learning
- Reading files line by line
- Applying operations to collections

## 5. For Loop with Conditions

````
for i in range(1, 11):
    if i % 2 == 0:
        print(i, "is even")
````

````
numbers = [5, 12, 7, 18, 3]
for num in numbers:
    if num > 10:
        print(num)
````

````
names = ["Ali", "Ahmed", "Sara"]
for name in names:
    if name.startswith("A"):
        print(name)
````

````
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

````
for i in range(1, 6):
    if i == 4:
        break
    print(i)
````
