## Dictionary

```python
student = {'name':'Ali','age':21}
student['age']
```

```python
student.update({'city':'Delhi'})
student
```

```python
student["name"]
```

```python
student.get("age")
student.get("marks", "Not Found")
```

```python
student["age"] = 22
student["city"] = "Delhi"
student
```

```python
student.keys()
```

```python
student.values()
```

```python
student.items()
```

```python
student.pop("city")
student
```
