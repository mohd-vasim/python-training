## 1. Introduction to Matplotlib
Matplotlib is the most commonly used Python library for data visualization.

Key features:
- Low-level control over plots
- Supports line plots, scatter, bar charts, histograms, subplots, etc.
- Works seamlessly with NumPy & pandas


```
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('default')
```

## 2. Basic Line Plot

```

x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.figure(figsize=(6,4))
plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

## 3. Customizing Lines and Markers

```

plt.figure(figsize=(6,4))
plt.plot(x, y, linestyle='--', marker='o')
plt.title("Line With Markers")
plt.show()
```

## 4. Scatter Plot

```

np.random.seed(0)
x = np.random.randn(100)
y = np.random.randn(100)
plt.figure(figsize=(6,4))
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 5. Bar Chart

```

categories = ["A", "B", "C", "D"]
values = [23, 45, 56, 12]
plt.figure(figsize=(6,4))
plt.bar(categories, values)
plt.title("Bar Chart Example")
plt.show()
```

## 6. Histogram

```

data = np.random.randn(1000)
plt.figure(figsize=(6,4))
plt.hist(data, bins=30)
plt.title("Histogram Example")
plt.show()
```

## 7. Subplots – Multiple Plots in One Figure

```

fig, axs = plt.subplots(1, 2, figsize=(10,4))
axs[0].plot(np.sin(x))
axs[0].set_title("Sine")

axs[1].plot(np.cos(x))
axs[1].set_title("Cosine")

plt.tight_layout()
plt.show()
```

## 8. Labels, Titles, Legends, Grid

```

plt.figure(figsize=(6,4))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("Trigonometric Functions")
plt.xlabel("Angle")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```

## 10. Pie Chart

```

sizes = [30, 20, 25, 25]
labels = ["Python", "Java", "C++", "JS"]
plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Programming Language Popularity")
plt.show()
```

## 11. Box Plot

```

plt.figure(figsize=(6,4))
plt.boxplot([np.random.randn(100), np.random.randn(100)+1])
plt.title("Box Plot Example")
plt.show()
```

## 12. Using Matplotlib with pandas

```

import pandas as pd
df = pd.DataFrame({
    "month": ["Jan","Feb","Mar","Apr","May"],
    "sales": [100, 120, 90, 150, 180]
})
df.plot(x="month", y="sales", kind="line", figsize=(6,4))
plt.title("Monthly Sales")
plt.show()
```
