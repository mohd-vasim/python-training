# Pandas – Core Concepts, Fundamentals, and Essential Methods

This notebook teaches pandas from scratch with detailed explanations, examples, and exercises.

## 1. What Is Pandas?
Pandas is the most widely used Python library for data analysis.

Key features:
- Provides `Series` (1D) and `DataFrame` (2D tabular data)
- Fast, optimized operations built on NumPy
- Easy data cleaning, transformation, joining, grouping, and slicing
- Supports CSV, Excel, SQL, JSON, Parquet, and more

```python
import pandas as pd
import numpy as np
pd.__version__
```

## 2. Pandas Series (1D)
A **Series** is a 1D labeled array, similar to a NumPy array but with labels.

```python

s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
s, s.index, s.values
```

### Important Attributes:

```python

print("Index:", s.index)
print("Values:", s.values)
print("Data type:", s.dtype)
print("Shape:", s.shape)
```

### Basic Operations

```python

print(s + 10)
print(s * 2)
print(s[s > 20])
```

## 3. DataFrame – The Core of Pandas
A **DataFrame** is a 2D, table-like structure with rows and columns.

```python

data = {
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [24, 30, 22, 28],
    "salary": [50000, 60000, 45000, 70000]
}
df = pd.DataFrame(data)
df
```

### 3.1 DataFrame Attributes

```python

df.head(), df.tail(), df.shape, df.columns, df.index, df.dtypes
```

## 5. Selecting Data (The Most Important Skill)

### 5.1 Selecting Columns

```python

df["name"]
df[["name", "salary"]]
```

### 5.2 Selecting Rows

```python

print(df.loc[0])        # by row label

print(df.iloc[0])       # by index position

df.iloc[1:3]     # slicing
```

### 5.3 Boolean Filtering

```python

df[df["age"] > 25]
df[(df["age"] > 25) & (df["salary"] > 55000)]
```

## 6. Sorting

```python

df.sort_values("salary", ascending=False)
```

## 7. Adding / Modifying Columns

```python

df["salary_k"] = df["salary"] / 1000
df
```

## 8. Dropping Columns & Rows

```python

df.drop(columns=["salary_k"], inplace=False) # columns
df.drop(index=[2], inplace=False) # rows
```

## 9. Handling Missing Values (Important in Real Projects)

```python

df2 = df.copy()
df2.loc[1, "salary"] = np.nan

df2.isna(), df2.fillna(0), df2.dropna()
```

## 10. GroupBy – Split, Apply, Combine

```python

data = {
    "team": ["A", "A", "B", "B", "C"],
    "score": [10, 15, 10, 20, 30]
}
gdf = pd.DataFrame(data)
gdf.groupby("team")["score"].mean()
```

## 11. Aggregations

```python

gdf.groupby("team").agg({
    "score": ["sum", "mean", "max", "count"]
})
```

## 12. Merging / Joining DataFrames

```python

left = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["A", "B", "C"]
})
right = pd.DataFrame({
    "id": [1, 2, 4],
    "score": [90, 80, 70]
})
left.merge(right, on="id", how="inner")
```

## 13. Pivot Tables

```python

pt = pd.DataFrame({
    "city": ["Hyd", "Hyd", "Delhi", "Delhi"],
    "year": [2023, 2024, 2023, 2024],
    "sales": [100, 150, 80, 130]
})
pt.pivot_table(values="sales", index="city", columns="year")
```

## 14. Apply / Lambda Functions (Powerful)

```python

df["age_category"] = df["age"].apply(lambda x: "young" if x < 25 else "adult")
df
```

## 15. Ranking

```python

df["salary_rank"] = df["salary"].rank(ascending=False)
df
```
